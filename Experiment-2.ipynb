```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv("./datasets/iris.csv")
iris.head()
```

```python
columns = iris.select_dtypes(include="number").columns[1:]
columns[1:]
```

```python
for column in columns:
    plt.figure(figsize=(10,6))
    plt.hist(iris[column])
    plt.title(column)
    plt.show()
```

```python
for column in columns:
    plt.figure(figsize=(10,6))
    plt.boxplot(iris[column])
    plt.title(column)
    plt.show()
```

```python
plt.figure(figsize=(10, 6))

iris[columns].boxplot()

plt.title("Boxplots of Iris Features")
plt.ylabel("Value")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

iris[columns].hist()

plt.title("Boxplots of Iris Features")
plt.ylabel("Value")
plt.xticks(rotation=20)
plt.show()
```

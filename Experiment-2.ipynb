```python
import pandas as pd
```

# Loading dataset

```python
df = pd.read_csv("./datasets/titanic.csv")
df.head()
```

# Describing the dataset

```python
df.describe()
```

# Null Count

```python
df.isnull().sum()
```

# preprocessing null data

```python
df["Age"] = df["Age"].fillna(df["Age"].median()) # median due to high variance
```

```python
df = df.drop("Cabin", axis=1) # majority data missing
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

# normalisation

```python
min_max_normilizer = lambda x : (x - x.min())/(x.max() - x.min())
```

```python
df[["Age", "Fare"]] = df[["Age", "Fare"]].apply(min_max_normilizer)
```
